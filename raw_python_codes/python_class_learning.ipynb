# **Przeciążanie operatorów - operator overloading**

```
* Dodawanie (+): __add__(self, other)

* Odejmowanie (-): __sub__(self, other)

* Mnożenie (*): __mul__(self, other)

* Dzielenie dokładne (/): __truediv__(self, other)

* Dzielenie całkowite (//): __floordiv__(self, other)

* Reszta z dzielenia / Modulo (%): __mod__(self, other)

* Potęgowanie (**): __pow__(self, other)
```

````python
class Wektor:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __str__(self):
    return f"Wektor({self.x}, {self.y})"

  def __add__(self, inny_wektor):
    if not isinstance(inny_wektor, Wektor):
      raise TypeError("Mozna dodac do siebie tylko 2 wektory!")

    nowe_x = self.x + inny_wektor.x
    nowe_y = self.y + inny_wektor.y

    return Wektor(nowe_x, nowe_y)

  def __sub__(self, inny):
    if not isinstance(inny, Wektor):
      raise TypeError("Mozna odjac od siebie tylko 2 wektory!")

    return Wektor(self.x - inny.x, self.y - inny.y)

  def __mul__(self, liczba):
    if not isinstance(liczba, (int, float)):
      raise TypeError("Mnozenie tylko przez liczbe")

    return Wektor(self.x * liczba, self.y * liczba)

  def __rmul__(self, liczba):
    if not isinstance(liczba, (int, float)):
      raise TypeError("Mnozenie tylko przez liczbe")

    return Wektor(liczba * self.x, liczba * self.y)
````

````python
w1 = Wektor(2, 3)
w2 = Wektor(5, 1)
````

````python
wynik = w1 + w2
````

````python
print(f"Wynik: {wynik}")
````

````python
print(w1, w2, wynik)
````

````python
wynik = w1 - w2
print(wynik)
````

````python
wynik = w1 * 2
print(wynik)
````

````python
wynik = 10 * w2
print(wynik)
````

# **Metody porównawcze**

```
* __eq__ (od equal): definiuje ==

* __ne__ (od not equal): definiuje !=

* __lt__ (od less than): definiuje <

* __le__ (od less or equal): definiuje <=

* __gt__ (od greater than): definiuje >

* __ge__ (od greater or equal): definiuje >=
```

````python
class Produkt:
  def __init__(self, nazwa, cena):
    self.nazwa = nazwa
    self.cena = cena

  def __repr__(self):
    return f"{self.nazwa}({self.cena}zł)"

  def __eq__(self, inny):
    if not isinstance(inny, Produkt):
      raise TypeError("Można porównywać tylko 2 produkty ze sobą.")

    return self.cena == inny.cena

  def __lt__(self, inny):
    if not isinstance(inny, Produkt):
      raise TypeError("Można porównywać tylko 2 produkty ze sobą.")

    return self.cena < inny.cena
````

````python
p1 = Produkt("Klawiatura", 150)
p2 = Produkt("Myszka", 80)
p3 = Produkt("Kabel", 80)
p4 = Produkt("Monitor", 800)
````

````python
print(f"Czy myszka kosztuje tyle samo co kabel?: {p2 == p3}")
````

````python
koszyk = [p1, p2, p3, p4]
````

````python
print(koszyk)
````

# **Enkapsulacja (hermetyzacja)**

````python
class Produkt:
  def __init__(self, nazwa: str, cena: int | float):
    self.nazwa = nazwa
    self.cena = cena

  def __repr__(self):
    return f"{self.nazwa}({self.cena}zł)"

  def __eq__(self, inny):
    if not isinstance(inny, Produkt):
      raise TypeError("Można porównywać tylko 2 produkty ze sobą.")

    return self.cena == inny.cena

  def __lt__(self, inny):
    if not isinstance(inny, Produkt):
      raise TypeError("Można porównywać tylko 2 produkty ze sobą.")

    return self.cena < inny.cena

  # getter
  @property
  def cena(self):
    print(f"Odczyt ceny z gettera: ")
    return self._cena

  # setter
  @cena.setter
  def cena(self, nowa_cena):
    print("Ustawienie nowej ceny: ")

    if not isinstance(nowa_cena, (int, float)):
      raise TypeError("Cena musi być liczbą.")

    if nowa_cena < 0:
      raise ValueError("Cena musi być dodatnia.")

    self._cena = nowa_cena
````

````python
p1.cena = 200
````

# Dekoratory

````python
import time

def zmierz_czas(funkcja):
  def wrapper(*args, **kwargs):
    start = time.time()

    wynik = funkcja(*args, **kwargs)

    koniec = time.time()

    print(f"Funkcja {funkcja.__name__} wykonała sie w {koniec - start:.4f} sekund/y.")

    return wynik

  return wrapper
````

````python
@zmierz_czas
def obliczenia(sekundy: int):
  print(f"Start obliczeń trwających {sekundy} sekund/y...")
  time.sleep(sekundy)
  return "Gotowe!"
````

````python
@zmierz_czas
def mnozenie(a: int, b: int):
  return a * b
````

````python
wynik1 = obliczenia(2)

print("Wynik 1: ", wynik1)
````

````python
wynik2 = mnozenie(10, 5)

print("Wynik 2: ", wynik2)
````

Wielokrotny dekorator

````python
def powtorz(n: int):
  def dekorator(funkcja):
    def wrapper(*args, **kwargs):
      wynik = None
      print(f"Powtarzanie funkcji {funkcja.__name__} {n} razy")

      for i in range(n):
        print(f"{i + 1}.: ", end="")
        wynik = funkcja(*args, **kwargs)

      return wynik

    return wrapper

  return dekorator
````

````python
@powtorz(n=3)
def powitanie(imie):
  print(f"Hello, {imie}!")
  return True
````

# Metody klasowe i statyczne

````python
class Produkt:
  def __init__(self, nazwa, cena):
    self.nazwa = nazwa
    self.cena = cena

  def __str__(self):
    return f"{self.nazwa} - {self.cena} zł"

  @classmethod
  def tekst(cls, tekst):
    nazwa, cena = tekst.split("-")

    return cls(nazwa, int(cena))

  @staticmethod
  def czy_cena_prawidlowa(cena):
    if cena < 0 or cena > 100000:
      return False
    return True
````

````python
p1 = Produkt("Myszka", 80)

p2 = Produkt.tekst("Klawiatura-150")
````

````python
print(p1, p2)
````

````python
cena_test = -100
````

# Metody fabryczne (factory methods)

````python
import json
````

````python
class ZbiorDanych:
  def __init__(self, dane: list):
    self.dane = dane
    print(f"Utworzony został zbiór z {len(self.dane)} rekordami.")

  @classmethod
  def z_pliku_json(cls, txt_json: str):
    dane_json = json.loads(txt_json)

    return cls(dane_json)

  @classmethod
  def z_pliku_csv(cls, txt_csv: str):
    lines = txt_csv.strip().split("\n")
    heads = lines[0].split(",")

    data = []
    for line in lines[1:]:
      values = line.split(",")
      data.append(dict(zip(heads, values)))

    return cls(data)
````

````python
def podziel_dane(a, b):
  try:
    wynik = a / b
  except ZeroDivisionError:
    print("Nie można dzielić przez zero!")
    return None
  except TypeError:
    print("Prawdopodobnie podano tekst zamiast liczb/y!")
    return None
  else:
    return f"Sukces - wynik: " + str(wynik)
  finally:
    print("\nKoniec operacji")
````

# Enkapsulacja + wyjątki

````python
class InvalidPriceError(Exception):
  pass
````

````python
class Produkt:
  def __init__(self, nazwa: str, start_cena: float):
    self.nazwa = nazwa
    self.cena = start_cena

  @property
  def cena(self) -> float:
    return self._cena

  @cena.setter
  def cena(self, nowa_cena: float) -> None:
    if nowa_cena < 0:
      raise InvalidPriceError(f"Cena nie może być ujemna - podano {nowa_cena}.")
    elif nowa_cena > 100000:
      raise InvalidPriceError(f"Cena wymaga autoryzacji, gdyż jest za duża.")
    else:
      self._cena = nowa_cena
````

````python
def aktualizacja_ceny(produkt: Produkt, nowa_cena: float):
  print(f"\nala-API: próba zmiany ceny na: {produkt.nazwa} - z {produkt.cena} na {nowa_cena}...")

  try:
    produkt.cena = nowa_cena
    print("Sukces!")
  except InvalidPriceError as e:
    print(f"Odrzucono: {e}")

  except Exception as e:
    print(f"Inny błąd: {e}")
````

````python
laptop = Produkt("Laptop1", 5000)

aktualizacja_ceny(laptop, 4500)
````

````python
aktualizacja_ceny(laptop, -100)
````

````python
aktualizacja_ceny(laptop, 200000)
````

````python
class Pracownik:
  def __init__(self, imie: str, pensja: float):
    self.imie = imie
    self.pensja = pensja

  def pracuj(self):
    print(f"{self.imie.capitalize()} wykonuje zadania.")

  def przedstawienie_sie(self):
    print(f"Nazywam się {self.imie.capitalize()}.\
    Moje zarobki to {self.pensja}")
````

Mixin

````python
class ExportJSONMixin:
  def export_json(self):
    return json.dumps(self.__dict__, ensure_ascii=False)

class LogowanieMixin:
  def logowanie(self, akcja):
    print(f"LOG: Wykonano: {akcja} w obiekcie {self.__class__.__name__}")

class Dokumnet:
  def __init__(self, tytul) -> None:
    self.tytul = tytul
````

````python
class Raport(LogowanieMixin, ExportJSONMixin, Dokumnet):
  def __init__(self, tytul, zawartosc):
    super().__init__(tytul)
    self.zawartosc = zawartosc
````

````python
r = Raport("Wyniki 1", "Sprzedaż w góre o 30%")

r.logowanie("Utworzenie raportu")
````

````python
dane_json = r.export_json()
````
